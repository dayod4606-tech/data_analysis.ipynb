```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")
df.head()
```

```python

print("Shape:", df.shape)
print()
print(df.dtypes)
print()
print(df.isnull().sum())
print()
print(df.duplicated().sum())
```

```python
df.describe()
```

```python
df["tip_pct"] = (df["tip"] / df["total_bill"]) * 100
df["tip_pct"].describe()
```

```python
import matplotlib.pyplot as plt

df.groupby("day")["tip_pct"].mean().plot(kind="bar", title="Average Tip % by Day")
plt.ylabel("Tip %")
plt.show()
```

```python
df.groupby("smoker")["tip_pct"].mean().plot(kind="bar", title="Average Tip % by Smoker Status", color=["#2a9d8f", "#e63946"])
plt.ylabel("Tip %")
plt.show()
```

```python
import seaborn as sns

sns.scatterplot(data=df, x="total_bill", y="tip", hue="time")
plt.title("Total Bill vs Tip Amount")
plt.show()
```

## Key Findings

- Average tip percentage is about 16.1% across all bills
- Friday had the highest average tip % (17.0%), Saturday had the lowest (15.3%)
- Smokers tipped slightly more on average than non-smokers (16.3% vs 15.9%) — a small gap, not a strong effect
- Tip amount scales roughly linearly with bill size — bigger bills, bigger tips
